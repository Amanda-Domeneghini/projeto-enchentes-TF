Script:

```python
import requests
import json
import pandas as pd

# Link da primeira API (Casas de Bomba EBAP):
url = "https://gis-smamus.portoalegre.rs.gov.br/server/rest/services/Hosted/Casas_de_Bomba___EBAP_210126/FeatureServer/0/query"
      # query no final pra conseguir acessar os dados e nao o HTML da pagina

# Parametros de consulta (campos de ArcGIS Server - queremos todos os dados):
parametros = {
    "where": "1=1", # Todos os dados
    "outFields": "*", # Todas as colunas
    "f": "json" # Formato json
}

resposta = requests.get(url, params=parametros)

print("Status da requisição da API Casas de Bomba EBAP:", resposta.status_code)

# Tratamento de erros: somente se a requisicao foi bem-sucedida
if resposta.status_code == 200:

    # Transforma os dados coletados da API em arquivo JSON:
    dados = resposta.json()
    
    # Salva o arquivo JSON localmente na pasta do projeto
    with open("../dados/originais/dadosapi_casasbombaebap.json", "w", encoding="utf-8") as f:
        json.dump(dados, f, ensure_ascii=False, indent=4)
        # Escreve arquivo com o nome especificado, utf-8 para aceitar acentos no nome
        
    print("Arquivo JSON salvo com sucesso!") # Se chegou ate aqui, deu certo

    # Teste para ver se funcionou (deve dar 24):
    print("A quantidade de casas de bomba eh de", len(dados["features"]))
else:
    print("Um erro inesperado ocorreu: ", resposta.text)
```
